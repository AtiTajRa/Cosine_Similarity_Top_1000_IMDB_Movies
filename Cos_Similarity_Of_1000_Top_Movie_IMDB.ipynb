# Libraries

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

# DataFrame

```python
df = pd.read_csv("imdb_top_1000.csv")
df = df.loc[: , ["Series_Title","Released_Year","Runtime","Genre","Director","Star1","Star2","Star3"]]
df["Combined_Features"] = df["Series_Title"] + " " + df["Released_Year"] + " " + df["Runtime"] + " " + df["Genre"] + " " + df["Director"] + " " + df["Star1"] +" " + df["Star2"] +" " + df["Star3"]
combined_df = df["Combined_Features"]
combined_df
```

# Vectorize


```python
#Initialization
vectorizer = CountVectorizer()
vectorized_df = vectorizer.fit_transform(combined_df)
```

# Cosine_Similarities

```python
similarities = cosine_similarity(vectorized_df)

def Similarity(index):
    score = similarities[index]
    top_5 = score.argsort()[-6:-1][::-1]
    return print(df["Series_Title"][top_5],score[top_5])
```
